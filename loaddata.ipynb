```python
import pandas as pd
import requests
import json

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('agriculture_dataset.csv')
print(df.columns) 
#note: crop type is object
df.head()
```

```python
plt.scatter(df['Rainfall'], df['Chlorophyll_Content'])
plt.xlabel('Rainfall')
plt.ylabel('Chlorophyll Content')
plt.suptitle('Relationship between Rainfall and Chlorophyll Content')
plt.gcf().set_size_inches(6, 3)
```

```python
plt.scatter(df['Soil_pH'], df['Chlorophyll_Content'])
plt.xlabel('Soil pH')
plt.ylabel('Chlorophyll Content')
plt.suptitle('Relationship between Soil pH and Chlorophyll Content')
plt.gcf().set_size_inches(6, 3)
```
